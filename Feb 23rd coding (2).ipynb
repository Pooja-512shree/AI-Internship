```python
import seaborn as sns
import matplotlib.pyplot as plt
#correaltion
corr =df.corr(numeric_only =True)

#heatmap
plt.figure(figsize=(10,7))
sns.heatmap(corr, annot=True,cmap='coolwarm',fmt='.2f',linewidth=0.5)
plt.title('Correlation Heatmap')
plt.show()
```

```python

X =df[['Avg. Area Income' , 'Avg. Area House Age', 

'Avg. Area Number of Rooms',

'Avg. Area Number of Bedrooms',

'Area Population']]

Y = df[['Price']]
```

```python
#train test split
from sklearn.model_selection import train_test_split

#split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.4,random_state=101)
```

```python
# 24th of feb class import linear regression
from sklearn.linear_model import LinearRegression
```

```python
lm=LinearRegression()
lm
```

```python
# train the model 
lm.fit(X_train,Y_train)
```

```python
#intercept
print(lm.intercept_)
```

```python
#co-efficients
coeff= pd.DataFrame(lm.coef_.T,X.columns,columns=['Coefficient'])
coeff
```

```python
# Predict 
predictions = lm.predict(X_test)
```

```python
plt.figure(figsize=(10,6))
sns.regplot(data=df,x='Avg. Area Income',y='Price',scatter_kws={'alpha':0.4},line_kws = {'color':'red'})
plt.title('Best Fit Line')
plt.show()
```

```python
from sklearn import metrics

import numpy as np

# Show the values of MAE, MSE, RMSE

print('MAE:', metrics.mean_absolute_error(Y_test, predictions))

print('MSE:', metrics.mean_squared_error(Y_test, predictions))

print('RMSE:', np.sqrt(metrics.mean_squared_error(Y_test, predictions)))
```

```python
new_house = [[190000, 16, 8, 4, 30000]]

predicted_val = lm.predict(new_house)

print(f"Predicted Price for this house: ${float(predicted_val[0]):,.2f}")

# or

print(f"Predicted Price for this house: ${predicted_val[0].item():,.2f}")
```

```python
intercept = lm.intercept_[0]

coeffs = lm.coef_.flatten()        # most people drop the .T — try without it first
print("The Secret Formula is:")
print(f"Price = ({coeffs[0]:.2f} * Income) +")
print(f"        ({coeffs[1]:.2f} * House Age) +")
print(f"        ({coeffs[2]:.2f} * Rooms) +")
print(f"        ({coeffs[3]:.2f} * Bedrooms) +")
print(f"        ({coeffs[4]:.2f} * Population) +")
print(f"        {intercept:.2f}") 
```
